```python
def student(name,rollno,age):# mahender
    print("my name is",name)
    print("my rollno is",rollno)
    print("my age is",age)
    # print("my branch is",branch)
    print()
```

```python
student("Mahender",1,21)
student(rollno=2,name="SIVA",age=19)# keyword
```

```python
def student(name,rollno,age,branch="CSE"):# default
    print("my name is",name)
    print("my rollno is",rollno)
    print("my age is",age)
    print("my branch is",branch)
    print()
```

```python

student("Mahender",1,21,"AI")
student(rollno=2,name="SIVA",age=19)
student("Shruthi",3,25)#default branch value
```

```python
def sum(*arr):#varble length
    s=0
    for i in arr:
        s+=i
    print(s)
    print(arr)
```

```python
sum(10,20,30)# required var len
sum(10,20,30,30,23,2,2,3,4,24,2,45,24)
sum(10)
```
