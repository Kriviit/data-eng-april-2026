### conditional statements
- if : if the condition gets true it will enter the block
- elif : if the previouse condtion gets false and the current condtion gets true it will enter the block
- else : if the previouse condtion gets false it will enter the block


```python
if False:
    print("I am if block")
```

```python
num=21
if num%2==0:#False
    print("EVEN")
else:
    print("ODD")
# if num%2!=0:
#     print("ODD")
```
